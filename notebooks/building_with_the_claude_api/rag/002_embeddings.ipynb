```python
# Client Setup
from dotenv import load_dotenv
import voyageai

load_dotenv()

client = voyageai.Client()
```

```python
# Chunk by section
import re


def chunk_by_section(document_text):
    pattern = r"\n## "
    return re.split(pattern, document_text)
```

```python
# Embedding Generation
def generate_embedding(text, model="voyage-3-large", input_type="query"):
    result = client.embed([text], model=model, input_type=input_type)

    return result.embeddings[0]
```

```python
with open("./report.md", "r") as f:
    text = f.read()

chunks = chunk_by_section(text)

# Could fail because VPN could be blocked
# API providers routinely block datacenter and VPN address ranges at the edge, 
# and the response you're seeing is the signature of that: 
# a minimal HTML 403 from the front-end proxy, no JSON, no API error code.
generate_embedding(chunks[0])
```
